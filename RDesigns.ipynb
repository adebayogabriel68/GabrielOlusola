```r
# Load the data into R
    data <- read.csv("/content/DesignBlock2CSV.csv")
```

```r
head(data)
```

```r
# Ensure correct column names
colnames(data) <- c("Block", "Designs", "Respose")
```

```r
# Convert grouping variable to factor
data$Designs <- as.factor(data$Designs)
```

```r
# One-way ANOVA
anova_result <- aov(Respose ~ Designs, data = data)
```

```r
# Check for and handle non-numeric values in Respose column
data$Respose <- as.numeric(data$Respose)

# Remove rows with NA values in the Respose column
data <- na.omit(data)

# One-way ANOVA
anova_result <- aov(Respose ~ Designs, data = data)

# Display ANOVA summary
summary(anova_result)
```

```r
# Check for missing values in each column
colSums(is.na(data))
```

```r
# Get the summary of the ANOVA
anova_summary <- summary(anova_result)

# Convert the summary to a data frame
anova_summary_df <- as.data.frame(anova_summary[[1]])

# Write the data frame to a CSV file
write.csv(anova_summary_df, "anova_summary.csv", row.names = TRUE)
```
